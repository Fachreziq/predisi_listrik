```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df['Global_active_power'] = pd.to_numeric(
    df['Global_active_power'],
    errors='coerce'
)

df['Global_active_power'].dropna().head(100).plot(figsize=(12,5))

plt.title("Grafik Konsumsi Listrik")
plt.xlabel("Data")
plt.ylabel("Global Active Power")

plt.show()
```
